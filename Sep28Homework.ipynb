### Opertors


### 1. Arithmetic Operators

```python
a = 15
b = 4
print(a + b) 
print(a - b)
print(a * b)
print(a / b) # Division operator 15 ÷ 4 = 3.75
print(a // b) # Floor division operator 15 ÷ 4 = 3.75
print(a % b) # Modulus operator 15 % 4 = 3
print(a ** b) # Power operator 15 * 15 * 15 * 15= 50625
```

### String + Arithmetic

```python
name = "Prasad"
print(name * 3)
print(name + "  " + name + "  " + name)
print()

s = "Python"
print(s + " Programming")
print(s * 3)
print()

name = "Prasad"
age = 23
print("Name:", name)
print("Age:", age)
```

### 2.Comparison Operators

```python
a = 10
b = 20
print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

```python
name1 = "Prasad"
name2 = "Prasad"

print(name1 == name2)
print(name1 != name2)
```

```python
age = 23
print(age >= 18)
print(age < 18)
```

### 3.Logical Operators

```python
Name = "Prasad"
age = 23
if age  >= 18:
    print(Name, "is eligible to vote.")
else:
    print(Name, "is not eligible to vote.")
```

```python
age = 23
city = "Phaltan"

print(age > 18 and city == "Phaltan")
```

```python
is_student = True
print(not is_student)
```

```python
x = 10
print(x > 20 or x < 15)
```
