#### 1. Even or Odd
You have given a number. Write a program to check whether the number is EVEN or ODD.
num = 12  ---> EVEN
num = 7   ---> ODD

```python
Num = int(input("Enter Your Number:"))
if Num % 2 == 0:
    print(Num , "The Number is Even")
else:
    print(Num , "The Number is Odd")
```

#### 7. Check Divisibility
Write a program to check whether a number is divisible by 2, 3, or both.
num = 12  ---> Divisible by both 2 and 3
num = 8   ---> Divisible by 2
num = 9   ---> Divisible by 3
num = 7   ---> Not divisible by 2 or 3

```python
num = int(input("Enter Your Number:"))

if num % 2 == 0 and num % 3 == 0:
    print(num, "is divisible by both 2 and 3.")

elif num % 2 == 0:
    print(num, "is divisible by 2 but not by 3.")

elif num % 3 == 0:
    print(num, "is divisible by 3 but not by 2.")
    
else:
    print(num, "is not divisible by either 2 or 3.")
```

#### 8. Calculate Grade
Given a student's marks, print the grade.
90 - 100  ---> A
75 - 89   ---> B
60 - 74   ---> C
40 - 59   ---> D
Below 40  ---> Fail

Example:
marks = 82  ---> Grade B

```python
Marks = int(input("Enter Your Marks:"))

if Marks >= 90 and Marks <= 100:
    print("Your Grade is A")

elif Marks >= 75 and Marks <= 89:
    print("Your Grade is B")

elif Marks >= 60 and Marks <= 74:
    print("Your Grade is C")

elif Marks >= 40 and Marks <= 59:
    print("Your Grade is D")

else:
    print("Fail")
```
